```python
import cv2
import matplotlib.pyplot as plt
```

```python
image_path = "C:\\Users\\evangelos.lampos\\ionianemsclone\\ionianems\\frontend\\src\\assets\\headerImage.jpg"  # Replace with your actual image file path
image = cv2.imread(image_path)

```

```python
if image is None:
    print("⚠️ Error: Could not load image. Check the file path.")
    exit()
```

```python
# Get image dimensions
height, width, _ = image.shape
print(f"Image Dimensions: {width}x{height}")
```

```python
# ✅ Define the bottom-right corner coordinates
crop_size = 150  # Size of the cropped region (adjust as needed)
```

```python
x1, y1 = width - crop_size, height - crop_size  # Bottom-right starting point
x2, y2 = width, height  # Bottom-right corner
```

```python
# ✅ Crop only the bottom-right corner
bottom_right_corner = image[y1:y2, x1:x2]
```

```python
# ✅ Convert BGR to RGB for correct color display
bottom_right_corner_rgb = cv2.cvtColor(bottom_right_corner, cv2.COLOR_BGR2RGB)
```

```python
# ✅ Display the cropped bottom-right corner using Matplotlib
plt.figure(figsize=(5, 5))
plt.imshow(bottom_right_corner_rgb)
plt.axis("off")  # Hide axes
plt.title("Bottom-Right Corner")
plt.show()
```

```python
# Select a pixel inside the cropped area
px_x, px_y = 10, 10  # Adjust based on the cropped image size
```

```python
# Get the BGR color at (px_x, px_y)
b, g, r = bottom_right_corner[px_y, px_x]
```

```python
# Print the color
print(f"Color at ({px_x}, {px_y}): BGR({b}, {g}, {r})")
print(f"RGB Format: ({r}, {g}, {b})")  # Converted to RGB
```
